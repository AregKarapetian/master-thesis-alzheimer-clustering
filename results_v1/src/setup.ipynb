# Setup

Every analysis notebook starts with `%run -i ../src/setup.ipynb`. This imports the common packages and runs the
code notebooks in order:

| Order | Notebook | Provides |
|---|---|---|
| 1 | `config.ipynb` | `cfg`: settings and the eight configurations |
| 2 | `data_utils.ipynb` | `load_analysis_data`, `build_design`, … |
| 3 | `psi.ipynb` | `compute_config_pairwise_separability`, `cliffs_delta`, … |
| 4 | `clustering_utils.ipynb` | `fit_configuration`, `bootstrap_partitions`, … |
| 5 | `evaluation_utils.ipynb` | `relabel_by`, `psi_column`, `format_psi_table`, … |
| 6 | `plot_utils.ipynb` | `cluster_profile`, `psi_heatmap`, … |

`-i` runs them in the notebook's own workspace, so their functions and `cfg` can be used directly. The paths are
relative, so the analysis notebooks have to be opened from the `analysis/` folder.
